### Matplotlib 


```python
import matplotlib.pyplot as plt
```

```python
monthly = df.groupby(df['order_date'].dt.to_period('M'))['sales'].sum()
```

```python
monthly.plot(kind='line', figsize=(10,5), title='Monthly Sales Trend')   #Monthly sales trend
```

```python
# Sales by region
df.groupby('region')['sales'].sum().sort_values().plot(kind='barh', title='Sales by Region')
```
